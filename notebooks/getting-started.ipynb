# Getting started with Baler Notebooks

This notebook runs in **your** Jupyter container. It talks to Baler over the Haystack API with the
API key in `BALER_API_KEY`, exactly like any other client. Baler never runs this code.

Docs: *Notebooks* and *baler_fn* in the Baler docs site (`/code/notebooks`, `/code/baler-fn`).

```python
import baler_fn

client = baler_fn.connect()   # BALER_URL, BALER_API_KEY, BALER_PROJECT from the container env
client.about()
```

## Read records

`read` takes any Haystack filter. `read_df` gives a DataFrame.

```python
points = client.read_df("point")
points.head()
```

## History

`his_read_df` returns a DataFrame indexed by UTC timestamp. Ranges are Haystack ranges: `today`, `yesterday`, `2026-11-01`, `2026-11-01,2026-11-07`.

```python
point_id = points["id"].iloc[0]
his = client.his_read_df(point_id, "today")
his.tail()
```

## Installed functions

Functions come from installed packages (entry-point group `baler.functions`). The sample package ships with this image.

```python
fns = baler_fn.list()
fns
```
